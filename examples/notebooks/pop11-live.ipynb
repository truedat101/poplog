# Pop-11, live

One persistent [Poplog](https://github.com/IoTone/poplog) session backs this whole notebook. Every `define` is **compiled to native machine code** as the cell runs, and everything you define survives to later cells — the 🌈 line after each cell is the real wall-clock cost of the run.

Install: `tools/jupyter/install-kernel.sh` (after the pop11 skill's one-line install).

```pop11
npr('hello from Pop-11');
2 ** 100 =>
```

## Define once, native forever

`fib` here is not interpreted: by the time `npr` runs, the procedure is machine code.

```pop11
define fib(n);
    if n < 2 then n else fib(n - 1) + fib(n - 2) endif
enddefine;
npr('fib(30) = ' sys_>< fib(30));
```

Cells share the session — `fib` is still there:

```pop11
vars i;
for i from 10 by 10 to 30 do
    npr(i sys_>< ' -> ' sys_>< fib(i));
endfor;
```

## Lists, pattern power, and the parts Lisp fans will recognise

```pop11
vars langs = [pop11 prolog lisp sml forth];
npr('Poplog speaks ' sys_>< length(langs) sys_>< ' languages');
applist(langs, procedure(w); npr('  - ' sys_>< w) endprocedure);
```

## Mishaps are conversational

An error doesn't kill the kernel — the session survives, state intact. (Run it: the next cell still knows `fib`.)

```pop11
;;; uncomment to see a mishap survive:
;;; undefined_procedure(42);
npr('fib survived everything: ' sys_>< fib(20));
```
